
# Model Development Results

## Experimental Setup

Two classifiers were evaluated: Logistic Regression and Random Forest.
The data was split into training and test partitions using an 80/20
stratified split, with random state 42. Five-fold cross-validation was
performed on the training partition.

Preprocessing was included within each scikit-learn pipeline to ensure
that imputation, encoding, and scaling were fitted independently within
each training fold.

## Cross-Validation Results

Logistic Regression achieved a mean ROC-AUC of 0.893, while Random
Forest achieved 0.894. Random Forest was selected using mean
cross-validation ROC-AUC, although the difference between the models
was very small.

## Held-Out Test Evaluation

Logistic Regression achieved 86.9% accuracy and 0.966 ROC-AUC on the
held-out test set. Random Forest achieved 88.5% accuracy and 0.946
ROC-AUC. Random Forest achieved 96.4% recall for the positive class.

The test set contains only 61 observations, so these results have
substantial uncertainty and should not be interpreted as proof that
one model is universally superior.

## Saved Artifacts

The selected fitted pipeline was saved as artifacts/best_model.joblib.
The cross-validation and test metrics were saved as
artifacts/training_metrics.json. These files support reproducibility
and subsequent deployment.
